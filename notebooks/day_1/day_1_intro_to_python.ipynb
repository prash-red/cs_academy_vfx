## 3. Math operators

## 4. Conditionals — `if` / `elif` / `else`

Sometimes code needs to make a decision. `if` runs a block of code only when a
condition is `True`. `elif` ("else if") checks another condition, and `else` catches
everything else.

Comparisons like `==` (equal to), `<`, `>`, `<=`, `>=` produce `True` or `False`.

````python
age = 15

if age < 13:
    print("You're a kid.")
elif age < 18:
    print("You're a teenager.")
else:
    print("You're an adult.")
````

## 5. Logical operators — `and`, `or`, `not`

Comparisons like `age < 18` produce a **boolean** — a value that's either `True` or
`False`. You can combine multiple booleans into one condition:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `and` | both sides must be `True` | `True and False` | `False` |
| `or` | at least one side must be `True` | `True or False` | `True` |
| `not` | flips `True`/`False` | `not True` | `False` |

This lets an `if` check more than one thing at once.

````python
age = 15
has_ticket = True

if age >= 13 and has_ticket:
    print("You can watch the movie.")
else:
    print("You cannot watch the movie.")

print(age >= 13)
print(has_ticket)
print(not has_ticket)
````

## 10. Watch it play

This next part is given to you — you don't need to understand every line yet, it
uses tools we haven't covered. Just run it once to define `play_animation()`, then
run the cell after that to watch your `frames` list play back as a video.

````python
import time
import matplotlib.pyplot as plt
from IPython.display import clear_output


def play_animation(frames, delay=0.1):
    """Plays a list of 9x9 (or any size) 0/1 grids back as a simple video."""
    for frame in frames:
        clear_output(wait=True)
        plt.figure(figsize=(3, 3))
        plt.imshow(frame, cmap="Greys", vmin=0, vmax=1)
        plt.axis("off")
        plt.show()
        time.sleep(delay)
````

````python
play_animation(frames)
````
